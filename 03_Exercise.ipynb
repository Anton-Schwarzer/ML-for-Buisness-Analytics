## Adjusting Data Types

Pandas cannot always recognise all data types correctly and automatically. We may therefore have to adjust the types of individual columns in the dataframe ourselves.

The Pandas library offers many different data types. The [Pandas type system](https://pandas.pydata.org/docs/reference/arrays.html) extends the [NumPy type hierarchy](https://numpy.org/doc/2.2/reference/arrays.scalars.html) and is quite complex. But for a start, it might be enough to know a few of the the most important types.

For the basic numerical types, these are:

| Basic Type | Description | Concrete types  |
|-|-|-|
| `bool` | booleans | `bool` |
| `int` | integers | `int8`, `int16`, `int32`, `int64` |
| `uint` | unsigned integers | `uint8`, `uint16`, `uint32`, `uint64` |
| `float` | floating point numbers | `float16`, `float32`, `float64`, `float128` |
| `complex` | complex numbers (f.e.: `[1.+0.j]`)<br>with imaginary part `.imag` and real part `.real` | `complex64`, `complex128`, `complex256` |

NumPy does not allow missing values for integer arrays. Therefore, an extra type is provided for integer columns with missing values: `Int` for signed nullable integers or `UInt` for unsigned nullable integers.

| Basic Type | Description | Concrete types  |
|-|-|-|
| `Int` | nullable integers | `Int8`, `Int16`, `Int32`, `Int64` |
| `UInt` | nullable unsigned integers | `UInt8`, `UInt16`, `UInt32`, `UInt64` |

Unlike in NumPy for strings the data type `object` is used.

````python
# Print df['dteday'].dtype to show the oject data type
print(df['dteday'].dtype)
````

Furthermore there are other data types, like `datetime64[ns]`.

````python
# use the pd.to_datetime function to convert the date
df['dteday'] = pd.to_datetime(df['dteday'])
print(df['dteday'].dtype)
````

Let's start by converting the integer columns. Due to `mnth`, `weathersit` and `cnt` containing missing values, we have to convert them to the nullable type `UInt` (i.e. `UInt16`).

````python
# convert the columns to UInt16
df[['mnth', 'weathersit', 'cnt']] = df[['mnth', 'weathersit', 'cnt']].astype('UInt16')
````

`windspeed` should be numeric, but when we try to convert it to a floating point number (`float`) we get a error:

````python
# try to change the type of windspeed to float
df['windspeed'] = df['windspeed'].astype(float)
````

<code><span style="color:red;">ValueError</span>: could not convert string to float: '-'</code>

In addition to missing values, the data set also contains values that have been marked as missing with a hyphen `-`. These must first be converted to `None` so that the column can then be converted to the floating point format `float`.

````python
#checkout the unique values
df['windspeed'].unique()
````

````python
# use the replace method to replace "-" with None and convert to float
df['windspeed'] = df['windspeed'].replace('-', None).astype(float)
````

All variables should now be of the correct data type.

````python
df.dtypes
````

Note that we could save a lot of space by converting to even slimmer datatypes. For example, the `season` columns could easily be stored in a `uint8` format instead of `int64`.

## Aggregation

Aggregation is a central part of statistic and important to get an overview over large datasets. Pandas offers a lot of built-in aggregation functions natively, but gives us also the possibility to perform custom aggregations.

The `df.describe` method returns important measurements for numeric columns:
* **count**: the number of non missing values
* Measures of central tendency: **mean** and **50%** percentile (median)
* Information on the distribution of the data: **min**imum and **max**imum, standart deviation **std**, and the **25%** and **75%** percentiles (first and third quantile)

````python
df.describe()
````

Mainly for categorical columns the `df[column].unique` and `df[column].value_counts` methods might be of interest.

````python
df['weathersit'].unique()
````

````python
df['weathersit'].value_counts()
````

Primarily for numerical columns, we have many different aggregation methods such as `min`, `max`, `mean`, `median` or `sum`.

````python
df['temp'].min(), df['temp'].mean(), df['temp'].median(), df['temp'].max()
````

````python
df['cnt'].sum()
````

`count` returns us the  number of non missing values.

````python
print(df['dteday'].count(), df['mnth'].count())
````

All of these methods can also be applied to multiple columns (i.e. the whole dataframe) if the aggregation method and data types are compatible.

````python
df.min()
````

The `aggregate` method can be used to apply custom aggregation functions.

````python
def smallesNonZero(column):
    return min(column[column > 0])

df['windspeed'].aggregate(smallesNonZero)
````

Also here we can aggregate multiple columns at once. Furthermore by passing them as a list, we can also apply multiple aggregations at once.

````python
df.select_dtypes('number').aggregate([smallesNonZero, 'count', 'mean', 'min', 'median', 'max', 'std'])
````

By passing a Python `dict` with columns as keys and aggregation functions as values, we can specify exactly which aggregation to apply to which column.

````python
df.aggregate({'windspeed':'mean', 'cnt':'median'})
````

If you want to summarise the data belonging to a particular attribute, you can use `df.groupby`. For example, you could calculate the total number of rental bikes by day:

````python
df.groupby('dteday')['cnt'].sum()
````

We can also use the `aggregate` method on a grouped dataframe.

````python
df.groupby('hr').aggregate({'windspeed':'mean', 'cnt':'sum'})
````

## Value-wise transformation: `map`

The `map` method can be passed a function that specifies a transformation of a single value. The function is then used by Pandas for all values in a column.

````python
def discretize(value):
    if value < 20: return 'cold'
    if value < 25: return 'nice'
    return 'warm'

df['atemp'].map(discretize).sample(5, random_state=3)
````

## Column- / row-wise transformation: `apply`

The `apply` method is similar to the `map` function, except that the function passed expects an entire column (default) or row (`axis=1`) as an argument.

````python
def isCyclingWeather(row):
    if row['weathersit'] == 1:
        return 15 < row['atemp'] < 25
    return False

df.apply(isCyclingWeather, axis=1).head(15)
````

## Changing values on a dataframe view

Suppose we want to use the dataset for a supervised machine learning task, we can use the variable `cnt` to train a model to predict the number of bikes rented. For the training and evaluation of our model, however, only those observations are of use for which there is a valid value in `cnt` (i.e. where the values are not missing). Let's select this subset and save it to a new variable.

````python
df_valid = df[~df['cnt'].isnull()]
df_valid
````

When we create a slice of a dataframe, a view is usually returned. This means that internally Pandas does not copy the whole data, but returns a wrapper object that operates on the original data. One of the advantages of this is that it is more memory efficient.

If we now try to add a new column to our view, we will get a warning.

````python
df_valid['test'] = 0
````

The same when we try to modify an existing column.

````python
df_valid['mnth'] = df_valid['dteday'].dt.month
````

To protect the user, Pandas doesn't allow you to change the values of views directly. Instead we can use the ``loc`` function as follows:

````python
df_valid.loc[:,'mnth'] = df_valid['dteday'].dt.month
df_valid.head()
````

By using the `loc` keyword, we explicitly allowed the view to be changed independently of the original dataframe. So the changes we made in `df_valid` do not affect `df`.

Use the `select_dtypes` and `apply` method to perform column-wise standardization of all columns of data type `number`.
